```python
def pad_tensor(tensor):
    try:
        while len(tensor) < 31:
            #tensor = tensor.reshape(-1,)
            tensor = torch.cat([tensor, tensor[-1:,:]]) #.repeat(1)
    except IndexError:
        tensor = tensor.reshape(-1,1)
        while len(tensor) < 31:
            tensor = torch.cat([tensor, tensor[-1:,:]]) #.repeat(1)
    return tensor
```
